Nesta aula, você começará a se acostumar com a linguagem Python, executando comandos simples que serão muito úteis futuramente.

Vamos começar com o comando mais usado pelos iniciantes de Python: escrever "Olá, Mundo!"

```python
#Exercício 1 - Imprimir "Olá, Mundo!"
print("Olá, Mundo!")
```

```python
#Exercício 2 - Imprimir "Olá, Mundo!" e "Bem-vindo ao Python!" em linhas separadas
print("Olá, Mundo!")
print("Bem-vindo ao Python!")
```

```python
#Exercício 3 - Imprimir "Olá, Mundo!" e "Bem-vindo ao Python!" na mesma linha
print("Olá, Mundo!", end=" ")
print("Bem-vindo ao Python!")   
```

Cuidados ao digitar códigos em Python:

1) Python é uma linguagem "case sensitive", ou seja, ela difere maiúsculas de minúsculas. Por exemplo, "print" é diferente de "Print". 

2) Aspas importam: sempre que abri-las, você tem que fechá-las. Por exemplo, o código não rodará se você digitar print("Olá, Mundo!) porque você não fechou as aspas após o ponto de exclamação.

3) Parênteses também são cruciais: assim como ocorre nas aspas, se você abrir um parêntesis, terá que fechá-lo. Do contrário, seu código emitirá uma mensagem de erro.

4) Espaço é tudo em Python! A linguagem se baseia conta os espaços em branco antes do início de cada linha para realizar as operações desejadas no código. Isso é o que chamados de identação. Em Python, seu código terá problemas se a identação estiver incorreta.

```python
#Exemplo da Importância da Identação
x = "(2+2) = 4"
if x = True:
    print("Isso é verdade!")
else:
    print("Isso é falso!")
```
