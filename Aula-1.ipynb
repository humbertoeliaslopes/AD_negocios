## <span style="color:blue"> Os 6 Princípios Cruciais da Sintaxe da Linguagem </span>

Nesta aula, você começará a se acostumar com a linguagem Python, aprendendo os princípios fundamentais da sintaxe.

### **<span style="color:green">PRINCÍPIO 1</span>**:  os comandos em Python são escritos em um documento denominado código ou script. Neste código, você pode escrever qualquer texto não executável usando o símbolo "<span style="color:green">#".

```python
# Exemplo 1 - Imprimir "Olá, Mundo!"
print("Olá, Mundo!")
```

### **<span style="color:green">PRINCÍPIO 2</span>**: Python é uma linguagem "case sensitive", ou seja, ela difere maiúsculas de minúsculas. Por exemplo, "print" é diferente de "Print".

```python
# Exemplo 2 - Diferença entre print e Print

print("Este é um exemplo de print correto.")
Print("Este é um exemplo de Print incorreto.")  # Isso causará um erro
```

### **<span style="color:green">PRINCÍPIO 3</span>**: aspas (duplas ou simples) importam: sempre que abri-las, você tem que fechá-las.

```python
# Exemplo 3 - Importância das Aspas

print("Aspas importam em Python!")
print('Aspas simples também funcionam!')

print("Aspas importam em Python!") # Nesta linha, eu me esqueci de fechar as aspas, o que resultará em um erro de sintaxe.
print('Aspas simples também funcionam!') # Nesta linha, eu me esqueci de fechar as aspas, o que resultará em um erro de sintaxe.
```

### **<span style="color:green">PRINCÍPIO 4</span>**: parênteses também são cruciais: assim como ocorre nas aspas, se você abrir um parêntese, terá que fechá-lo. Do contrário, seu código emitirá uma mensagem de erro.

```python
# Exemplo 4 - Importância dos Parênteses

print("Aspas importam em Python!") # Os parênteses e as aspas são essenciais para a sintaxe correta do comando print.

print("Aspas importam em Python!" # Nesta linha, eu me esqueci de fechar o parêntese, o que resultará em um erro de sintaxe.
```

### **<span style="color:green">PRINCÍPIO 5</span>**: espaço é tudo em Python! A linguagem se baseia conta os espaços em branco antes do início de cada linha para realizar as operações desejadas no código. Isso é o que chamados de identação. Em Python, seu código terá problemas se a identação estiver incorreta.

```python
#Exemplo 5 - Importância da Indentação

#Cenário 1: Indentação Correta

# Definindo as variáveis para o teste
senha_correta = "python123"
senha_digitada = "senha_errada"

print("--- Cenário 1: Indentação Correta ---")
if senha_digitada == senha_correta:
    print("Acesso permitido! Bem-vindo.")
else:
    print("Senha incorreta!")
    print("ALERTA: Conta bloqueada por segurança!")  # Esta linha DEPENDE do 'else'
```

O que acontece no Cenário 1? Como a linha do "ALERTA" está alinhada (indentada) para dentro do else, ela só vai rodar se a senha estiver errada. O output está correto.

*<span style="color:red">O Perigo da Indentação Errada (Bug Semântico)</span>*

Agora, veja o que acontece se movermos a última linha apenas quatro espaços para a esquerda, tirando-a de dentro do bloco else:

```python
# Cenário 2: Indentação Errada (Bug)

print("\n--- Cenário 2: Indentação Errada (Bug) ---")
# Vamos testar com a senha CORRETA agora
senha_digitada = "python123"

if senha_digitada == senha_correta:
    print("Acesso permitido! Bem-vindo.")
else:
    print("Senha incorreta!")

print("ALERTA: Conta bloqueada por segurança!")  # Ops! Ficou fora do 'else'
```

💡 ***Regra de ouro em Python***: Todo código que estiver recuado para a direita pertence ao bloco que está logo acima dele (geralmente terminando com :). Quando você volta o texto para a esquerda, você está dizendo ao Python: "Esse bloco acabou, volte para o fluxo normal".

***Dicas para Iniciantes:***

1. Padronização: A convenção oficial do Python (PEP 8) dita que você deve usar 4 espaços para cada nível de indentação. A recomendação é usar a barra de espaços, não a tecla TAB.

2. Erros de Sintaxe: Se você esquecer de indentar uma linha que deveria estar dentro de um if, o Python vai parar o programa e exibir um erro chamado IndentationError.

3. Erros de Lógica: Como vimos no Cenário 2, se você indentar errado mas a sintaxe for válida, o Python vai rodar o código, mas o programa fará coisas malucas.

Voltando aos princípios da sintaxe do Python:

### **<span style="color:green">PRINCÍPIO 6</span>**: como vimos, a identação é fundamental. No entanto, espaços dentro de uma mesma linha são irrelevantes para rodar o código. Eles, porém, podem melhorar muito a visualização das linhas.

```python
# Exemplo 6 - Irrelevância dos Espaços em Branco na Mesma Linha

x=1+2
print(x)

x = 1 + 2
print(x)    

x          =        1       +    2
print(x)
```

### **<span style="color:green">PRINCÍPIO 7</span>**: em Python, o fim da linha usualmente indica o fim do comando.

```python
# Exemplo 7 - Final do Comando

ponto_medio = 5 # Criamos uma variável chamada 'ponto_medio' e atribuímos o valor 5 a ela. O comando termina aqui, mesmo que haja um comentário depois.
print(ponto_medio) # Imprime o valor da variável 'ponto_medio', que é 5. O comentário depois do comando não afeta a execução do código.
```

Alternativamente, o final de um comando poder ser indicado assim:

```python
# 1. Usando a barra invertida para continuar o comando na próxima linha

x = 1 + 2 + 3 + 4 +\
5 + 6 + 7 + 8
print(x)
```

```python
# 2. Usando parênteses para continuar o comando na próxima linha

x = (1 + 2 + 3 + 4 +
5 + 6 + 7 + 8)
print(x)

# O uso de parênteses é geralmente preferível, pois torna o código mais legível e evita erros comuns associados ao uso da barra invertida.
```
