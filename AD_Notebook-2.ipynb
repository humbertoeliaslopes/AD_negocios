## <span style="color:blue"> 1) Tabela com Distribuição de Frequências </span>

Como estudamos no Notebook 1, a distribuição de frequências é um resumo tabular de dados mostrando o número (frequência) de observações em cada uma das diversas classes não sobrepostas. (Anderson et. al., 2021, p. 29). Essa definição vale tanto para variáveis categorizadas como quantitativas.

Neste caso, porém, precisamos ter mais cuidado porque dados quantitativos podem facilmente se sobrepor. Consequentemente, não conseguiríamos registrar o comportamento desses dados por meio de distribuições de frequência porque a sobreposição contrariaria as condições impostas na definição acima. Logo, precisamos manipular os dados quantitativos - "manipulação", nesta acepção, não implica valoração negativa - para que consigamos agrupá-los adequadamente em classes que comporão a distribuição de frequências.

De acordo com Anderson et al. (2021, p. 35), os procedimentos corretos para atingirmos esse objetivo são:

1) Determinar o número de classes não sobrepostas.
2) Determinar a amplitude de cada classe.
3) Determinar os limites da classe.

### **<span style="color:green">Na Prática 2.1</span>**:  Criando uma distribuição para os tempos necessários para auditar dados no final do ano. 

O livro-texto da disciplina apresenta um problema prático por meio da Tabela 2.4. Nela, estão os dados relativos ao tempo, em dias, necessário para concluir auditorias de fim de ano para uma amostra de 20 clientes da Sanderson & Clifford, uma pequena empresa de contabilidade pública. Os dados são os seguintes:

12 14 19 18
15 15 18 17
20 27 22 23
22 21 33 28
14 18 16 13


Comecemos definindo o número de clases não sobrepostas (Procedimento 1).

A diretriz geral é agrupar os dados quantitativos utilizando entre 5 e 20 classes, de tal forma que não se tenha classes com muitos elementos e outras, com poucos. Nesse primeiro exemplo, temos um conjunto com 20 dados quantitativos, o que nos permite criar 5 classes contendo 4 elementos em cada.

Agora que definimos o número de classes, determinemos sua amplitude (Procedimento 2).

Para isso, usaremos a fórmula da amplitude da classe aproximada, expressa por:

$$
\text{Amplitude da classe aproximada} = \frac{\text{Maior valor dos dados} - \text{Menor valor dos dados}}{\text{Número de classes}}
$$

No conjunto de dados do exemplo temos que o maior valor é 33 e o menor, 12. Logo:
$$
\text{Amplitude da classe aproximada} = \frac{\text{33} - \text{12}}{\text{5}}= 4,2
$$

Vamos arredondar esse valor para cima, definindo, assim, uma amplitude da classe de 5. 

Finalmente, podemos criar as 5 classes determinando seus limites inferior e superior (Procedimento 3).

Dessa forma, as classes serão:

Classe 1: de 10 a 14 dias.
Classe 2: de 15 a 19 dias.
Classe 3: de 20 a 24 dias.
Classe 4: de 25 a 29 dias.
Classe 5: de 30 a 34 dias.

Repare que a diferença entre os limites de cada classe é sempre fixa, representada pelo valor da amplitude que, neste exemplo, é de 5 dias.

**ATENÇÃO**

Muitas pessoas se confundem sobre isso. Afinal, se calcularmos a diferença entre 14 e 10 (o primeiro intervalo de classe), o resultado algébrico será 4 dias, e não 5. Mas essa diferença é calculada considerando-se o desde o menor valor até o maior. Assim, temos, no primeiro intervalo:

10, 11, 12, 13 e 14: cinco valores, ou seja, o tamanho do intervalo de classe calculado.

O código em Python para criar a tabela com essa distribuição é o seguinte:

```python
import numpy as np
import pandas as pd

# 1. Base de dados da amostra (n = 20)
tempos_auditoria = [
    12,
    14,
    19,
    18,
    15,
    15,
    18,
    17,
    20,
    27,
    22,
    23,
    22,
    21,
    33,
    28,
    14,
    18,
    16,
    13,
]

serie_dados = pd.Series(tempos_auditoria, name="Tempo")

# 2. Definição das classes inteiras fechadas
bins = [9, 14, 19, 24, 29, 34]
rotulos_classes = ["10 a 14", "15 a 19", "20 a 24", "25 a 29", "30 a 34"]

serie_classes = pd.cut(
    serie_dados,
    bins=bins,
    labels=rotulos_classes,
    right=True,
)

# 3. Construção do DataFrame de frequências
tabela = serie_classes.value_counts(sort=False).reset_index()
tabela.columns = ["Tempo de auditoria (dias)", "Frequência simples"]

total_amostra = len(serie_dados)
tabela["Frequência relativa"] = tabela["Frequência simples"] / total_amostra
tabela["Frequência percentual (%)"] = tabela["Frequência relativa"] * 100

# 4. Adição da linha de totais
linha_total = pd.DataFrame(
    {
        "Tempo de auditoria (dias)": ["Total"],
        "Frequência simples": [total_amostra],
        "Frequência relativa": [1.00],
        "Frequência percentual (%)": [100.0],
    }
)

tabela_completa = pd.concat([tabela, linha_total], ignore_index=True)

# 5. Formatação numérica (Padrão brasileiro: vírgula decimal)
tabela_exibicao = tabela_completa.copy()
tabela_exibicao["Frequência relativa"] = (
    tabela_exibicao["Frequência relativa"]
    .map(lambda x: f"{x:.2f}".replace(".", ","))
)
tabela_exibicao["Frequência percentual (%)"] = (
    tabela_exibicao["Frequência percentual (%)"]
    .map(lambda x: f"{x:.1f}".replace(".", ","))
)

# 6. Aplicação dos Estilos CSS conforme ABNT / IBGE
# - Sem bordas verticais
# - Linha superior e inferior do cabeçalho
# - Linha inferior delimitando o final do corpo da tabela
# - Destaque tipográfico sutil no total
tabela_estilizada = (
    tabela_exibicao.style
    .set_caption("<b>Tabela 1</b> – Distribuição de frequências do tempo de auditoria contábil (em dias)")
    .set_table_styles([
        {"selector": "caption", "props": [("caption-side", "top"), ("text-align", "left"), ("color", "#000000"), ("font-size", "11pt"), ("margin-bottom", "8px")]},
        {"selector": "table", "props": [("border-collapse", "collapse"), ("font-family", "Arial, sans-serif"), ("font-size", "10pt"), ("width", "100%"), ("color", "#000000")]},
        {"selector": "th", "props": [("border-top", "1.5pt solid black"), ("border-bottom", "1pt solid black"), ("padding", "6px 12px"), ("text-align", "center"), ("background-color", "transparent")]},
        {"selector": "td", "props": [("border", "none"), ("padding", "5px 12px"), ("text-align", "center")]},
        {"selector": "tbody tr:last-child", "props": [("border-top", "1pt solid black"), ("border-bottom", "1.5pt solid black"), ("font-weight", "bold")]}
    ])
    .hide(axis="index")
)

# Visualização direta no Jupyter Notebook
tabela_estilizada
```

Uma outra estatística que pode ser importante em determinados casos é o ponto médio da classe. Ele consiste no valor intermediário entre os limites inferior e superior de classe. Assim, por exemplo:

$$M_i = \frac{\text{Limite Inferior} + \text{Limite Superior}}{2}$$

Para a classe de **10 a 14** dias de tempo de auditoria:

* **Limite Inferior ($LI$):** $10$
* **Limite Superior ($LS$):** $14$

$$M_1 = \frac{10 + 14}{2} = \frac{24}{2} = 12\text{ dias}$$

## <span style="color:blue"> 2) Diagrama de Pontos (Dot Plot) </span>

O diagrama de pontos é uma opção para registrar o comportamento de dados quantitativos. Na prática, porém, ele é pouco usado porque outro tipo de gráfico - o histograma de frequências - exibe ainda mais informações do que ele.

Apenas a título de exemplo, veja como seria um diagrama de pontos dos dados que estamos utilizando até agora neste Notebook.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 1. Base de dados da amostra: tempo de auditoria (em dias), n = 20
tempos_auditoria = [
    12,
    14,
    19,
    18,
    15,
    15,
    18,
    17,
    20,
    27,
    22,
    23,
    22,
    21,
    33,
    28,
    14,
    18,
    16,
    13,
]

# 2. Algoritmo de empilhamento vertical para o gráfico de pontos
# Conta as repetições de cada valor para definir as coordenadas verticais (1, 2, 3, ...)
valores_unicos, contagens = np.unique(tempos_auditoria, return_counts=True)

coord_x = []
coord_y = []

for valor, contagem in zip(valores_unicos, contagens):
    for nivel in range(1, contagem + 1):
        coord_x.append(valor)
        coord_y.append(nivel)

# 3. Construção do gráfico de pontos com padrão visual acadêmico
plt.figure(figsize=(10, 4))

# Plotagem dos pontos empilhados
plt.scatter(
    coord_x,
    coord_y,
    color="#1f77b4",
    edgecolor="#0b3c5d",
    linewidth=1.2,
    s=110,
    zorder=3,
    label="Auditoria individual",
)

# Inclusão de linha de referência operacional (exemplo de limite contratual de 20 dias)
plt.axvline(
    x=20,
    color="#d62728",
    linestyle="--",
    linewidth=1.3,
    label="Ponto de controle (20 dias)",
    zorder=2,
)

# 4. Ajustes de eixos, títulos e escalas
plt.title(
    "Gráfico de Pontos: Tempo de Auditoria Contábil (em dias)",
    fontsize=12,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Tempo de Auditoria (dias)", fontsize=10, labelpad=8)
plt.ylabel("Frequência de Ocorrências", fontsize=10, labelpad=8)

# Configuração dos limites e marcações dos eixos
todos_os_dias = np.arange(10, 36, 1)
marcas_principais = np.arange(10, 36, 2)

plt.xticks(marcas_principais)
plt.yticks(range(1, max(coord_y) + 2))
plt.xlim(9, 35)
plt.ylim(0.4, max(coord_y) + 0.8)

# Grade horizontal e vertical discreta para guiar a leitura
plt.grid(axis="x", linestyle=":", color="#cccccc", alpha=0.8, zorder=0)
plt.grid(axis="y", linestyle=":", color="#e0e0e0", alpha=0.6, zorder=0)

# Remoção de bordas desnecessárias (estilo limpo)
plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)

plt.legend(frameon=True, facecolor="#ffffff", edgecolor="#cccccc", fontsize=9)
plt.tight_layout()
plt.show()
```

## <span style="color:blue"> 3) Histograma de Frequências </span>

O histograma de frequências é uma das formas mais comuns de representar o comportamento de dados quantitativos. Ele é especialmente importante em análises inferenciais, nas quais o comportamento dos dados amostrais é o foco do estudo. A seguir, você tem um exemplo de código Python para criar um histograma com os dados que usamos até agora.

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Conjunto de dados da amostra (n = 20 auditorias em dias)
tempos_auditoria = [
    12,
    14,
    19,
    18,
    15,
    15,
    18,
    17,
    20,
    27,
    22,
    23,
    22,
    21,
    33,
    28,
    14,
    18,
    16,
    13,
]

# 2. Definição das fronteiras reais de classe e pontos médios
# Fronteiras contínuas garantem barras adjacentes sem lacunas artificiais
fronteiras = [9.5, 14.5, 19.5, 24.5, 29.5, 34.5]
pontos_medios = [12, 17, 22, 27, 32]
rotulos_classes = ["10 a 14", "15 a 19", "20 a 24", "25 a 29", "30 a 34"]

# 3. Configuração da figura com foco em clareza editorial
plt.figure(figsize=(9, 5))

# Plotagem do histograma com barras contíguas (rwidth=1.0)
contagens, bordas, barras = plt.hist(
    tempos_auditoria,
    bins=fronteiras,
    edgecolor="#1c3d5a",
    color="#4a90e2",
    linewidth=1.2,
    rwidth=1.0,
    zorder=3,
)

# 4. Inclusão dos valores de frequência absoluta no topo de cada barra
for i in range(len(contagens)):
    plt.text(
        pontos_medios[i],
        contagens[i] + 0.2,
        f"fi = {int(contagens[i])}\n({(contagens[i]/len(tempos_auditoria))*100:.0f}%)",
        ha="center",
        va="bottom",
        fontsize=9,
        fontweight="bold",
        color="#222222",
    )

# Linha de referência gerencial (prazo-meta padrão de 20 dias)
plt.axvline(
    x=19.5,
    color="#d62728",
    linestyle="--",
    linewidth=1.5,
    label="Fronteira Crítica de Controle (20 dias)",
    zorder=4,
)

# 5. Formatação de eixos e rótulos
plt.title(
    "Histograma: Tempo de Execução de Auditorias Contábeis",
    fontsize=12,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Tempo de Auditoria (dias agrupados em classes)", fontsize=10, labelpad=8)
plt.ylabel("Frequência Absoluta (Número de Auditorias)", fontsize=10, labelpad=8)

# Posicionamento das marcações no eixo X exatamente nos pontos médios das classes
plt.xticks(pontos_medios, rotulos_classes)
plt.yticks(range(0, int(max(contagens)) + 3))
plt.ylim(0, max(contagens) + 1.8)
plt.xlim(9.5, 34.5)

# Linhas de grade sutis apenas no eixo Y para leitura de altura
plt.grid(axis="y", linestyle=":", color="#cccccc", alpha=0.7, zorder=0)

# Remoção de bordas superiores e laterais direitas
plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)

plt.legend(frameon=True, facecolor="#ffffff", edgecolor="#cccccc", loc="upper right")
plt.tight_layout()
plt.show()
```

O histograma de frequências pode ser combinado com uma curva de frequências acumuladas, aumentando o nível de informação que é apresentada para o analista de dados. Essa curva acumulativa é chamada de ogiva de Galton ou polígono de frequências. Veja um exemplo.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 1. Conjunto de dados da amostra: tempo de auditoria contábil (em dias), n = 20
tempos_auditoria = [
    12,
    14,
    19,
    18,
    15,
    15,
    18,
    17,
    20,
    27,
    22,
    23,
    22,
    21,
    33,
    28,
    14,
    18,
    16,
    13,
]

n_total = len(tempos_auditoria)

# 2. Parâmetros das classes
# Fronteiras reais contínuas para contiguidade física das barras
fronteiras = [9.5, 14.5, 19.5, 24.5, 29.5, 34.5]
rotulos_classes = ["10 a 14", "15 a 19", "20 a 24", "25 a 29", "30 a 34"]
pontos_medios = [12, 17, 22, 27, 32]

# Contagem das frequências absolutas por classe
contagens, _ = np.histogram(tempos_auditoria, bins=fronteiras)

# Frequências percentuais acumuladas para os pontos da ogiva
# Ponto inicial: (fronteira_inf_classe_1, 0%)
x_ogiva = fronteiras
y_ogiva = [0.0] + list((np.cumsum(contagens) / n_total) * 100)

# 3. Configuração da figura e eixos duplos
fig, ax1 = plt.subplots(figsize=(10, 5.5))

# Eixo Primário (Esquerda): Histograma de Frequências Absolutas
barras = ax1.hist(
    tempos_auditoria,
    bins=fronteiras,
    color="#4a90e2",
    edgecolor="#1c3d5a",
    linewidth=1.2,
    rwidth=1.0,
    alpha=0.75,
    zorder=2,
    label="Frequência Simples (nº de auditorias)",
)

ax1.set_xlabel(
    "Tempo de Execução de Auditoria (dias agrupados em classes)",
    fontsize=10,
    labelpad=10,
)
ax1.set_ylabel(
    "Frequência Simples (Auditorias)",
    fontsize=10,
    color="#1c3d5a",
    labelpad=8,
)
ax1.set_xticks(pontos_medios)
ax1.set_xticklabels(rotulos_classes, fontsize=9)
ax1.set_ylim(0, max(contagens) + 2.5)
ax1.tick_params(axis="y", labelcolor="#1c3d5a")
ax1.grid(axis="y", linestyle=":", color="#cccccc", alpha=0.6, zorder=0)

# Rótulos no topo de cada barra do histograma
for pm, cont in zip(pontos_medios, contagens):
    ax1.text(
        pm,
        cont + 0.25,
        f"{int(cont)}",
        ha="center",
        va="bottom",
        fontsize=9,
        fontweight="bold",
        color="#1c3d5a",
    )

# Eixo Secundário (Direita): Curva de Frequência Acumulada (Ogiva)
ax2 = ax1.twinx()
linha_ogiva = ax2.plot(
    x_ogiva,
    y_ogiva,
    color="#d95f02",
    linewidth=2.2,
    marker="o",
    markersize=6,
    zorder=4,
    label="Frequência Percentual Acumulada (%)",
)

ax2.set_ylabel(
    "Frequência Percentual Acumulada (%)",
    fontsize=10,
    color="#d95f02",
    labelpad=8,
)
ax2.set_ylim(0, 105)
ax2.set_yticks(np.arange(0, 101, 20))
ax2.tick_params(axis="y", labelcolor="#d95f02")

# Rótulos percentuais em cada ponto da ogiva (a partir do 1º limite superior)
for x, y in zip(x_ogiva[1:], y_ogiva[1:]):
    ax2.annotate(
        f"{y:.0f}%",
        xy=(x, y),
        xytext=(0, 7),
        textcoords="offset points",
        ha="center",
        fontsize=8.5,
        fontweight="bold",
        color="#d95f02",
    )

# Linha de referência gerencial para prazo operacional de 20 dias (fronteira 19,5)
ax1.axvline(
    x=19.5,
    color="#333333",
    linestyle="--",
    linewidth=1.2,
    zorder=3,
)
ax2.annotate(
    "Marco de 20 dias úteis\n(60% concluídos)",
    xy=(19.5, 60),
    xytext=(21.5, 45),
    arrowprops=dict(arrowstyle="->", color="#333333", lw=1),
    fontsize=8.5,
    bbox=dict(
        boxstyle="round,pad=0.4",
        facecolor="#fdfdfd",
        edgecolor="#cccccc",
        alpha=0.9,
    ),
)

# 4. Título e Legenda Unificada
plt.title(
    "Distribuição e Acumulação do Tempo de Auditoria Contábil (n = 20)",
    fontsize=11,
    fontweight="bold",
    pad=15,
)

# Consolidação dos itens de legenda dos dois eixos
legenda_elementos = [barras[2][0], linha_ogiva[0]]
legenda_rotulos = [b.get_label() for b in legenda_elementos]
ax1.legend(
    legenda_elementos,
    legenda_rotulos,
    loc="upper left",
    frameon=True,
    facecolor="#ffffff",
    edgecolor="#dddddd",
    fontsize=8.5,
)

# Ajuste visual das bordas superiores
ax1.spines["top"].set_visible(False)
ax2.spines["top"].set_visible(False)

plt.tight_layout()
plt.show()
```

## <span style="color:blue"> 4) Diagrama Ramo-e-Folhas (Stem-and-Leaf Plot) </span>

Esse diagrama é útil quando queremos exibir simultaneamente a ordem de classificação e o formato de uma distribuição de dados. 

O código Python a seguir mostra um exemplo desse diagrama.

```python
import numpy as np
import pandas as pd

# 1. Conjunto de dados da amostra (n = 20 auditorias em dias)
tempos_auditoria = [
    12,
    14,
    19,
    18,
    15,
    15,
    18,
    17,
    20,
    27,
    22,
    23,
    22,
    21,
    33,
    28,
    14,
    18,
    16,
    13,
]

# 2. Algoritmo didático para construção do diagrama de ramo e folhas
def diagrama_ramo_folhas(dados, unidade_folha=1):
    dados_ordenados = sorted(dados)
    min_ramo = min(dados_ordenados) // 10
    max_ramo = max(dados_ordenados) // 10

    # Inicialização dos ramos contínuos (garante ramos intermediários mesmo sem dados)
    estrutura_ramos = {ramo: [] for ramo in range(min_ramo, max_ramo + 1)}

    for valor in dados_ordenados:
        ramo = valor // 10
        folha = valor % 10
        estrutura_ramos[ramo].append(str(folha))

    # Impressão formatada no padrão acadêmico
    print("=" * 48)
    print("DIAGRAMA DE RAMO E FOLHAS: TEMPO DE AUDITORIA")
    print(f"Tamanho da amostra (n) = {len(dados)}")
    print(f"Unidade da Folha (Leaf Unit) = {unidade_folha}")
    print("=" * 48)
    print(f"{'Ramo':>4} | {'Folhas'}")
    print("-" * 48)

    for ramo in sorted(estrutura_ramos.keys()):
        folhas_formatadas = " ".join(estrutura_ramos[ramo])
        print(f"{ramo:4d} | {folhas_formatadas}")

    print("=" * 48)


# Execução do diagrama
diagrama_ramo_folhas(tempos_auditoria)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ==============================================================================
# 1. CARREGAMENTO E PREPARAÇÃO DOS DADOS
# ==============================================================================
url_github = "https://raw.githubusercontent.com/humbertoeliaslopes/AD_negocios/1bbd8810336e902587b32c997edd95cac31dd23e/BestPayingDegrees.xlsx"

try:
    df = pd.read_excel(url_github)
except Exception:
    try:
        df = pd.read_excel("BestPayingDegrees.xlsx")
    except Exception:
        # Recomposição fiel dos 20 cursos com maior remuneração intermediária
        dados = {
            "Curso": [
                "Engenharia Aeroespacial", "Engenharia Química", "Ciência da Computação",
                "Engenharia Nuclear", "Matemática Aplicada", "Engenharia Biomédica",
                "Física", "Engenharia da Computação", "Economia", "Engenharia de Petróleo",
                "Engenharia Elétrica", "Estatística", "Filosofia", "Finanças",
                "Engenharia Mecânica", "Sistemas de Informação", "Geologia",
                "Construção Civil", "Engenharia Civil", "Ciência Atuarial"
            ],
            "Salario_Inicial": [
                59400, 64800, 59800, 63900, 54300, 54800, 50300, 61200, 48800, 93000,
                60800, 50000, 39900, 47500, 58300, 49300, 43500, 51600, 53500, 54100
            ],
            "Salario_Intermediario": [
                108000, 108000, 102000, 104000, 101000, 101000, 97300, 87700, 97800, 157000,
                104000, 93400, 81200, 88300, 97400, 87100, 84200, 87400, 88100, 112000
            ],
            "Aumento_Pct": [
                82, 67, 71, 63, 86, 84, 93, 43, 100, 69,
                71, 87, 104, 86, 67, 77, 94, 69, 65, 107
            ]
        }
        df = pd.DataFrame(dados)

col_pct = [c for c in df.columns if "aumento" in c.lower() or "pct" in c.lower() or "%" in str(c).lower()][0]
total_obs = len(df)

# ==============================================================================
# 2. ITEM 1: TABELA DE DISTRIBUIÇÃO DE FREQUÊNCIAS (ABNT/IBGE + STORYTELLING)
# ==============================================================================
bins = list(range(40, 120, 10))  # Amplitude de classe = 10
rotulos = [f"[{bins[i]}, {bins[i+1]})" for i in range(len(bins) - 2)]
rotulos.append(f"[{bins[-2]}, {bins[-1]}]")

df["Classe"] = pd.cut(df[col_pct], bins=bins, right=False, include_lowest=True)
fi = df["Classe"].value_counts(sort=False).values

tabela_abnt = pd.DataFrame({
    "Aumento Percentual (%)": rotulos,
    "Ponto Médio (xi)": [(bins[i] + bins[i+1]) / 2 for i in range(len(bins) - 1)],
    "Frequência Absoluta (fi)": fi
})

tabela_abnt["Frequência Relativa (fr)"] = (tabela_abnt["Frequência Absoluta (fi)"] / total_obs).round(4)
tabela_abnt["Frequência Percentual (f%)"] = (tabela_abnt["Frequência Relativa (fr)"] * 100).round(2)
tabela_abnt["Frequência Acumulada (Fi)"] = tabela_abnt["Frequência Absoluta (fi)"].cumsum()
tabela_abnt["Freq. Percentual Acum. (F%)"] = (tabela_abnt["Frequência Acumulada (Fi)"] / total_obs * 100).round(2)

# Formatação terminal aberta padrão ABNT
larguras = [28, 18, 25, 25, 28, 25, 28]
L_TOTAL = sum(larguras)
colunas = list(tabela_abnt.columns)

print("=" * L_TOTAL)
print("Tabela 1: Concentração Bimodal: 55% das Carreiras de Topo Têm Saltos Salariais entre 60%-70% ou 80%-90%")
print("Distribuição do aumento percentual entre o salário inicial e intermediário dos 20 cursos mais bem pagos (America.EDU)")
print("=" * L_TOTAL)

cabecalho = "".join(f"{col:<{larguras[i]}}" if i == 0 else f"{col:>{larguras[i]}}" for i, col in enumerate(colunas))
print(cabecalho)
print("-" * L_TOTAL)

for _, row in tabela_abnt.iterrows():
    intervalo_str = str(row[colunas[0]])
    marcador = "*" if intervalo_str == "[60, 70)" else " "
    linha = (
        f"{marcador + intervalo_str:<{larguras[0]}}"
        f"{row[colunas[1]]:>{larguras[1]}.1f}"
        f"{int(row[colunas[2]]):>{larguras[2]}}"
        f"{row[colunas[3]]:>{larguras[3]}.4f}"
        f"{row[colunas[4]]:>{larguras[4]}.2f}"
        f"{int(row[colunas[5]]):>{larguras[5]}}"
        f"{row[colunas[6]]:>{larguras[6]}.2f}"
    )
    print(linha)

print("-" * L_TOTAL)
linha_total = (
    f"{'Total':<{larguras[0]}}"
    f"{'':>{larguras[1]}}"
    f"{total_obs:>{larguras[2]}}"
    f"{1.0000:>{larguras[3]}.4f}"
    f"{100.00:>{larguras[4]}.2f}"
    f"{'':>{larguras[5]}}"
    f"{'':>{larguras[6]}}"
)
print(linha_total)
print("=" * L_TOTAL)
print("Fonte: America.EDU (29 de agosto de 2013), adaptado por Anderson et al.")
print("Nota: Linha assinalada com asterisco (*) demarca o intervalo modal da amostra.\n")

# ==============================================================================
# 3. ITEM 2: HISTOGRAMA COM STORYTELLING DE DADOS (KNAFFLIC)
# ==============================================================================
fig, ax = plt.subplots(figsize=(10.5, 5.8), dpi=300)
fig.patch.set_facecolor("#FFFFFF")
ax.set_facecolor("#FFFFFF")

pontos_medios = tabela_abnt["Ponto Médio (xi)"].values
frequencias = tabela_abnt["Frequência Absoluta (fi)"].values

# Hierarquia intencional de cores
cores = []
for f in frequencias:
    if f == 6:
        cores.append("#1A365D")  # Azul marinho de destaque (moda)
    elif f == 5:
        cores.append("#2B6CB0")  # Azul intermediário (segundo polo)
    elif f > 0:
        cores.append("#718096")  # Cinza ardósia neutro
    else:
        cores.append("#E2E8F0")  # Cinza claro para barra nula

barras = ax.bar(
    pontos_medios,
    frequencias,
    width=10,
    color=cores,
    edgecolor="#FFFFFF",
    linewidth=1.2,
    align="center"
)

# Rotulagem direta acima das barras (decluttering)
for barra, freq in zip(barras, frequencias):
    if freq > 0:
        pct_val = (freq / total_obs) * 100
        ax.text(
            barra.get_x() + barra.get_width() / 2,
            barra.get_height() + 0.18,
            f"{int(freq)} ({pct_val:.0f}%)",
            ha="center",
            va="bottom",
            fontsize=9.5,
            fontweight="bold",
            color="#2D3748"
        )

# Remoção de ruído visual (decluttering)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_visible(False)
ax.spines["bottom"].set_color("#718096")
ax.spines["bottom"].set_linewidth(1.1)

ax.yaxis.set_visible(False)
ax.set_xticks(bins)
ax.set_xticklabels([f"{b}%" for b in bins], fontsize=9, color="#4A5568")
ax.tick_params(axis="x", length=4, color="#718096")
ax.set_xlim(35, 115)
ax.set_ylim(0, 8.2)

# Título ativo e contexto (Storytelling Knafflic)
fig.text(
    0.12, 0.95,
    "Evolução Salarial Apresenta Padrão Bimodal: Engenharias Concentram Ganhos Moderados e Áreas Teóricas Dobram de Valor",
    fontsize=11.5, fontweight="bold", color="#1A202C"
)
fig.text(
    0.12, 0.91,
    "Distribuição do aumento percentual entre o salário inicial e intermediário dos 20 cursos com maior remuneração de carreira (America.EDU)",
    fontsize=9, color="#718096"
)

# Anotações narrativas guiadas
ax.annotate(
    "Pico modal (30% dos cursos):\nEngenharias Tradicionais\n(piso inicial já elevado)",
    xy=(65, 6),
    xytext=(52, 7.0),
    arrowprops=dict(arrowstyle="->", color="#1A365D", lw=1.2),
    fontsize=8.5,
    fontweight="bold",
    color="#1A365D",
    ha="center"
)

ax.annotate(
    "Segundo polo (25%):\nFinanças, Estatística e Matemática",
    xy=(85, 5),
    xytext=(85, 6.4),
    arrowprops=dict(arrowstyle="->", color="#2B6CB0", lw=1.2),
    fontsize=8.5,
    fontweight="bold",
    color="#2B6CB0",
    ha="center"
)

ax.annotate(
    "Salto superior a 100%:\nFilosofia, Economia\ne Ciência Atuarial",
    xy=(105, 3),
    xytext=(105, 4.6),
    arrowprops=dict(arrowstyle="->", color="#4A5568", lw=1.2),
    fontsize=8.5,
    color="#4A5568",
    ha="center"
)

plt.subplots_adjust(top=0.85)
plt.savefig("histograma_best_paying_degrees_storytelling.png", dpi=300, bbox_inches="tight")
plt.show()

# ==============================================================================
# 4. ITEM 3: APRESENTAÇÃO RAMO-E-FOLHAS COM STORYTELLING
# ==============================================================================
def ramo_e_folhas_storytelling(series):
    print("=" * 90)
    print("DIAGRAMA RAMO-E-FOLHAS: DINÂMICA DE ACELERAÇÃO SALARIAL AO LONGO DA CARREIRA")
    print("Unidade do Ramo: 10% | Unidade da Folha: 1% | Amostra: n = 20 cursos")
    print("=" * 90)
    
    valores = sorted(int(round(v)) for v in series)
    ramos = {}
    for v in valores:
        s = v // 10
        f = v % 10
        ramos.setdefault(s, []).append(f)
        
    perfil_analitico = {
        4: "Ponto de partida isolado (Eng. Computação)",
        5: "[Intervalo Vazio: ausência de graduações]",
        6: "MODA: Engenharias Tradicionais (Petróleo, Mecânica)",
        7: "Tecnologia e Infraestrutura (Computação, Civil)",
        8: "2º POLO: Finanças, Estatística e Matemática",
        9: "Ciências Exatas Fundamentais (Física, Geologia)",
        10: "ARRANCADA (>100%): Filosofia, Economia e Atuária"
    }
    
    print(f"{'Ramo':>4} | {'Folhas':<26} | {'Freq.':>5} | Perfil Analítico / Cursos Representados")
    print("-" * 90)
    
    min_ramo, max_ramo = min(bins) // 10, (max(bins) - 10) // 10
    for r in range(min_ramo, max_ramo + 1):
        folhas_lista = ramos.get(r, [])
        folhas_str = "  ".join(str(x) for x in folhas_lista)
        comentario = perfil_analitico.get(r, "")
        print(f"{r:>4} | {folhas_str:<26} | {len(folhas_lista):>5} | {comentario}")
        
    print("-" * 90)
    print(f"Total: n = {len(valores)} observações | Média = {np.mean(valores):.1f}% | Mediana = {np.median(valores):.1f}%")
    print("=" * 90 + "\n")

ramo_e_folhas_storytelling(df[col_pct])
```
