## A situação da empresa

A **Grão Mineiro** é uma rede fictícia de cafeterias com 12 lojas em Belo Horizonte. Em reuniões com os gerentes, a diretoria ouviu uma queixa recorrente: nos horários de pico, as filas no caixa são longas e os atendentes não têm tempo de oferecer produtos adicionais, como um pão de queijo junto com o café ou uma fatia de bolo.

Para enfrentar o problema, a empresa instalou em **outubro** um **totem de autoatendimento** em cada uma das 12 lojas. No totem, o próprio cliente monta o pedido, e a tela sugere itens complementares antes do pagamento. A diretora de operações, Camila Rezende, quer saber se o investimento valeu a pena. Ela pediu ao time de análise de dados que comparasse o **ticket médio** (valor médio gasto por pedido, em reais) de cada loja em **setembro**, último mês sem o totem, e em **outubro**, primeiro mês com o totem.

Informações financeiras fornecidas pela controladoria:

* o aluguel de cada totem custa **R\$ 1.500,00 por mês** por loja;
* cada loja atende, em média, **2.000 pedidos por mês**;
* a **margem de contribuição** dos produtos vendidos é de **65%**, ou seja, de cada R\$ 1,00 a mais vendido sobram R\$ 0,65 para cobrir custos fixos e gerar lucro.

O contrato de aluguel dos totens vence no fim de novembro, e Camila precisa decidir se o renova para as 12 lojas.
