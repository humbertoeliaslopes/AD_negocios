## A situação da empresa

O **Café Serrano** é uma rede fictícia de cafeterias, com 30 lojas na Região Metropolitana de Belo Horizonte. A diretora de marketing, Patrícia Andrade, quer lançar um programa de fidelidade em toda a rede e recebeu três propostas:

* **cashback**: o cliente recebe de volta 5% do valor gasto, para usar na compra seguinte;
* **pontos**: cada real gasto vale um ponto, e os pontos são trocados por produtos;
* **desconto progressivo**: o desconto cresce com o número de visitas no mês.

Para decidir, o Café Serrano fez um teste durante um mês. Dezenove lojas de porte semelhante foram **sorteadas** para os três programas: 6 lojas com cashback, 7 com pontos e 6 com desconto progressivo. No fim do mês, a equipe registrou o **faturamento mensal de cada loja**, em milhares de reais.

Informações fornecidas pela controladoria:

* a **margem de contribuição** dos produtos é de **30%**, ou seja, de cada R\$ 1,00 vendido sobram R\$ 0,30 para cobrir custos fixos e gerar lucro;
* o **custo mensal de cada programa, por loja** (benefícios concedidos aos clientes e operação da plataforma), é de **R\$ 3.500,00** para o cashback, **R\$ 1.200,00** para os pontos e **R\$ 2.000,00** para o desconto progressivo.

Patrícia precisa apresentar à diretoria qual programa deve ser adotado nas 30 lojas.
